```python
import os

import pandas as pd

meta = pd.read_csv("metadata.csv")
```

```python
# Print the number of items

print("Number of items:", len(meta))
```

```python
# Check that every path in the file column exists, and that every file in data appears exactly once in the file column

files_in_folder = []
for name in os.listdir("data"):
    path = "data/" + name
    if not name.startswith(".") and os.path.isfile(path):
        files_in_folder.append(path)

listed = list(meta["file"])

# Every path in the metadata should be a file in the folder
for path in listed:
    if path not in files_in_folder:
        print("In metadata.csv, but no file:", path)

# Every file in the folder should appear exactly once in the metadata
for path in files_in_folder:
    if listed.count(path) != 1:
        print("Not listed exactly once in metadata.csv:", path)

print("File check finished")
```

```python
# Check that all IDs are unique

print("Repeated IDs:", meta["id"].duplicated().sum())
```

```python
# Count the missing values in each column

meta.isna().sum()
```

```python
# Summarise the dataset by space shuttle

meta["space_shuttle_name"].value_counts()
```

```python
# Summarise the dataset by mission

meta["mission"].value_counts()
```

```python
# Summarise the dataset by decade published

decades = []
for date in meta["date_published"].dropna():
    year = int(str(date)[:4])
    decades.append(year // 10 * 10)

pd.Series(decades).value_counts().sort_index()
```
